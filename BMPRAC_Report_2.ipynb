# 1. Summary

This study investigated the relationship between depression severity and glycaemic health outcomes using data from NHANES spanning 2005–2016. Specifically, we examined whether higher scores on the PHQ-9, a measure of depressive symptoms, were associated with elevated levels of fasting glucose, HbA1c and 2-hour Oral Glucose Tolerance Test (OGTT) values, which are all biomarkers for glucose metabolism.

We used a merged dataset across six NHANES cycles to perform descriptive statistics, data visualisation and hypothesis testing such as independent t-tests and multiple linear regression. Our sample included adults aged 20–80 years from the U.S. and it was filtered to remove incomplete data resulting in a final sample size of 9154.

Findings from non-parametric tests displayed that all biomarkers significantly differed from one another, highlighting their different physiological roles. Fasting glucose corresponds to baseline blood sugar levels, HbA1c tracks long-term glucose exposure and OGTT measures the body's ability to process glucose after a meal.

However, when comparing the biomarkers across depression severity categories using the Kruskal-Wallis test, no consistent differences were observed.

Stratified analysis by gender, age and race did not provide strong evidence for depression as contributor to diabetic risk.

Furthermore, multiple linear regression models confirmed that insulin resistance (HOMA-IR) and age were the strongest and most consistent predictors of glycaemic outcomes across all three biomarkers.

Unexpected findings included the lack of statistical significance for BMI in the fasting glucose model, and the unusual pattern where individuals with less than an hour per day of sedentary time had higher glucose levels. This could potentially be due to uncontrolled variables or misclassification. PHQ-9 scores remained statistically non-significant, reaffirming earlier findings.

Since NHANES is a cross-sectional study, causal relationships between depression and diabetes cannot be determined. Moreover, the use of self-reported data may cause errors due to potential bias.

```python
# import relevant libraries for analysis
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats
import seaborn as sns
from scipy.stats import shapiro
import statsmodels.api as sm
!pip install scikit-posthocs

# define years and cycles for merging data
years = ["2005", "2007", "2009", "2011", "2013", "2015"]
cycles = ["D", "E", "F", "G", "H", "I"]
```

The code below is used to import and merge all data from the respective variables from NHANES.


---


At this step:
- for demographic data, the dataset filters out respondents aged less than 20 and more than 80
- for physical activity and PHQ-9 data, respondents who responded with 'Don't know' or 'Refused' for any of the 9 questions were filtered out  
- for physical activity data, the code `PAD590` is used from 2005 to 2010 while the code `PAQ710` is used from 2011 to 2016, for the purposes of this analysis, the code `PAD590` is used for all physical activity data
  - For the 2007 and 2009 cycles, the question on how many hours they spent watching TV/videos on average in the past 30 days was only asked to children aged 2 - 11.
  - Hence, we took 52% of the minutes spent sedentary (`PAD680`) by the participants and converted it to hours.
  - According to Matthews *et al.* (2021), about 52% of the time spent sedentary by Americans adults is spent on watching TV and videos. Therefore, only 52% of the minutes spent sedentary is used when convert the proxy data (used for the 2007 and 2009 cycles) to hours spent watching TV/videos
- for fasting glucose and fasting insulin data, fasting insulin data is found in a separate file (with the code INS) for the 2013-2014 and 2015-2016 cycles
- negligible small numbers found in the data after importing (e.g `5.397605e-79`) were converted to 0

```python
# Helper function to add year column
def append_cycle_year(df_list, year_list):
    for df, year in zip(df_list, year_list):
        df["YEAR"] = int(year)
    return df_list

# Helper function to load and process NHANES data
def load_nhanes_data(file_code, usecols):
    urls = [f"https://wwwn.cdc.gov/Nchs/Data/Nhanes/Public/{y}/DataFiles/{file_code}_{c}.xpt" for y, c in zip(years, cycles)]
    dfs = [pd.read_sas(url) for url in urls]
    dfs = append_cycle_year(dfs, years)
    df = pd.concat(dfs, ignore_index=True)
    return df[usecols + ["YEAR"]]

# --- Demographic Data ---
df_demo = load_nhanes_data("DEMO", ["SEQN", "RIAGENDR", "RIDAGEYR", "RIDRETH1"])
df_demo = df_demo[(df_demo["RIDAGEYR"] >= 20) & (df_demo["RIDAGEYR"] <= 80)]
df_demo = df_demo.astype({"SEQN": int, "RIAGENDR": int, "RIDAGEYR": int, "RIDRETH1": int})
print("Demographic data:")
print(df_demo.head())
print(df_demo.shape)

# --- BMI Data ---
df_bmi = load_nhanes_data("BMX", ["SEQN", "BMXBMI"])
df_bmi = df_bmi.astype({"SEQN": int, "BMXBMI": float})
print("BMI data:")
print(df_bmi.head())
print(df_bmi.shape)

# --- Depression (PHQ-9) Data ---
dpq_urls = [f"https://wwwn.cdc.gov/Nchs/Data/Nhanes/Public/{y}/DataFiles/DPQ_{c}.xpt" for y, c in zip(years, cycles)]
dpq_dfs = [pd.read_sas(url) for url in dpq_urls]
dpq_dfs = append_cycle_year(dpq_dfs, years)
df_phq = pd.concat(dpq_dfs, ignore_index=True)
dpq_vars = [f"DPQ0{i:02d}" for i in range(10, 100, 10)]
df_phq = df_phq[["SEQN"] + dpq_vars + ["YEAR"]]
df_phq[dpq_vars] = df_phq[dpq_vars].apply(pd.to_numeric, errors='coerce')
threshold = 1e-5
df_phq[dpq_vars] = df_phq[dpq_vars].map(lambda x: 0 if isinstance(x, float) and abs(x) < threshold else x)
df_phq = df_phq[~df_phq[dpq_vars].isin([7, 9]).any(axis=1)]
df_phq = df_phq.dropna(subset=dpq_vars)
df_phq['PHQ9_total'] = df_phq[dpq_vars].sum(axis=1)
df_phq = df_phq.astype({"SEQN": int, "PHQ9_total": int})
print("PHQ-9 data with total score:")
print(df_phq.head())
print(df_phq.shape)

# --- Physical Activity (Sedentary Time) Merging Across Cycles ---
activity_frames = []
for year, cycle in zip(years, cycles):
    url = f"https://wwwn.cdc.gov/Nchs/Data/Nhanes/Public/{year}/DataFiles/PAQ_{cycle}.xpt"
    df_cycle = pd.read_sas(url)
    # For 2005: Use PAD590 (TV hours)
    if year == "2005" and "PAD590" in df_cycle.columns:
        df_sub = df_cycle[["SEQN", "PAD590"]].copy()
        df_sub = df_sub[df_sub["PAD590"].isin([0, 1, 2, 3, 4, 5, 6])]
    # For 2011, 2013, 2015: Use PAQ710 (TV hours)
    elif year in ["2011", "2013", "2015"] and "PAQ710" in df_cycle.columns:
        df_sub = df_cycle[["SEQN", "PAQ710"]].copy()
        df_sub = df_sub[df_sub["PAQ710"].isin([0, 1, 2, 3, 4, 5, 6, 8])]
        df_sub["PAQ710"] = df_sub["PAQ710"].replace(8, 6)  # '8' means 'None'
        df_sub.rename(columns={"PAQ710": "PAD590"}, inplace=True)
    # For 2007 & 2009: Use PAD680 (sedentary minutes), convert to hours + classify
    elif year in ["2007", "2009"] and "PAD680" in df_cycle.columns:
        df_sub = df_cycle[["SEQN", "PAD680"]].copy()
        df_sub = df_sub[df_sub["PAD680"].notna()]
        df_sub["PAD590"] = df_sub["PAD680"] * 0.52 / 60.0  # Convert to hours
        def classify_sedentary(hours):
            if hours < 0.25:
                return 6  # None
            elif hours < 1:
                return 0
            elif hours < 2:
                return 1
            elif hours < 3:
                return 2
            elif hours < 4:
                return 3
            elif hours < 5:
                return 4
            else:
                return 5
        df_sub["PAD590"] = df_sub["PAD590"].apply(classify_sedentary)
    else:
        print(f"Skipped {year} — no matching column found")
        continue
    # Clean + append
    df_sub["YEAR"] = int(year)
    df_sub = df_sub.astype({"SEQN": int, "PAD590": int})
    activity_frames.append(df_sub[["SEQN", "YEAR", "PAD590"]])
# Combine everything
df_pa = pd.concat(activity_frames, ignore_index=True)
print("Combined Physical Activity Data:")
print(df_pa.head())
print(df_pa["YEAR"].value_counts().sort_index())
print(df_pa.shape)

# --- Fasting Glucose and Insulin Data (with separate INS files for 2013 & 2015 cycles) ---
gluins_list = []
for y, c in zip(years, cycles):
    # Load glucose data
    glu_url = f"https://wwwn.cdc.gov/Nchs/Data/Nhanes/Public/{y}/DataFiles/GLU_{c}.xpt"
    df_glu = pd.read_sas(glu_url, index=None)
    df_glu = df_glu[["SEQN", "LBDGLUSI"]].copy()
    # Load insulin data from correct source
    if y in ["2013", "2015"]:
        ins_url = f"https://wwwn.cdc.gov/Nchs/Data/Nhanes/Public/{y}/DataFiles/INS_{c}.xpt"
    else:
        ins_url = glu_url  # insulin also in GLU file for earlier cycles
    df_ins = pd.read_sas(ins_url, index=None)
    df_ins = df_ins[["SEQN", "LBXIN"]].copy()
    # Merge glucose and insulin data
    df_cycle = df_glu.merge(df_ins, on="SEQN", how="inner")
    df_cycle["YEAR"] = int(y)
    gluins_list.append(df_cycle)
# Combine all into one dataframe
df_gluins = pd.concat(gluins_list, ignore_index=True)
df_gluins = df_gluins.astype({"SEQN": int, "LBDGLUSI": float, "LBXIN": float, "YEAR": int})
# Calculate HOMA-IR
df_gluins["HOMA_IR"] = (df_gluins["LBDGLUSI"] * df_gluins["LBXIN"]) / 22.5
df_gluins = df_gluins.astype({"SEQN": int, "LBDGLUSI": float, "LBXIN": float, "YEAR": int,"HOMA_IR":float})
print("Glucose, Insulin, and HOMA-IR data:")
print(df_gluins.head())
print(df_gluins.shape)

# --- HbA1c Data ---
df_hba1c = load_nhanes_data("GHB", ["SEQN", "LBXGH"])
df_hba1c = df_hba1c.astype({"SEQN": int, "LBXGH": float})
print("HbA1c data:")
print(df_hba1c.head())
print(df_hba1c.shape)

# --- OGTT Glucose Data ---
df_ogtt = load_nhanes_data("OGTT", ["SEQN", "LBDGLTSI"])
df_ogtt = df_ogtt.astype({"SEQN": int, "LBDGLTSI": float})
print("OGTT Glucose data:")
print(df_ogtt.head())
print(df_ogtt.shape)
```

# 5. Statistical Analysis

## Hypothesis testing

From report 1, we found that all of our variables were not normally distributed, even after removing outliers, through the Shapiro Wilk's test. Hence, non-parametric statistical tests will be used in this section.

### Comparison between diabetic biomakers
This section compares the diabetic biomakers with each other, to see if there is any statistically significant difference between them.

As we are comparing 3 biomaker values that were collected from the same individual, it will be appropriate to do a repeated measure ANOVA. However, since the data is not normally distributed, the non-parametric version of the test, the Friedman test, is done.

Hypotheses to be used:

Null hypothesis: All 3 diabetic biomarkers are not statistically different from each other.

Alternative hypotheses: At least 1 diabetic biomarkers are statistically different from the other biomarkers.

```python
# Subset only complete cases across all 3 biomarkers
biomarker_df = df_clean[["SEQN", "LBDGLUSI", "LBXGH", "LBDGLTSI"]].dropna()

# Melt to long format
melted = biomarker_df.melt(id_vars="SEQN",
                           value_vars=["LBDGLUSI", "LBXGH", "LBDGLTSI"],
                           var_name="Biomarker",
                           value_name="Value")
```

```python
from scipy.stats import friedmanchisquare

# Pivot to wide format by SEQN
pivot_df = melted.pivot(index="SEQN", columns="Biomarker", values="Value")

# Run Friedman test
stat, p = friedmanchisquare(
    pivot_df["LBDGLUSI"],
    pivot_df["LBXGH"],
    pivot_df["LBDGLTSI"]
)

print(f"Friedman test statistic = {stat:.4f}, p-value = {p:.4f}")
```

**Interpretation of results**

As the p-value of the test is < 0.05, at least one of the biomakers have a significantly different mean that the rest.

In order to further investigate, a post-hoc test using Wilcoxon's test is done. Wilcoxon's test is the non-parametric version of the paired t-test that compares 2 sets of non-normally distributed data.

```python
from scipy.stats import wilcoxon
from statsmodels.stats.multitest import multipletests

pairs = [
    ("LBDGLUSI", "LBXGH"),
    ("LBDGLUSI", "LBDGLTSI"),
    ("LBXGH", "LBDGLTSI")
]

p_vals = []
for a, b in pairs:
    stat, p = wilcoxon(pivot_df[a], pivot_df[b])
    print(f"Wilcoxon {a} vs {b}: p = {p:.4f}")
    p_vals.append(p)

# Bonferroni correction
adjusted = multipletests(p_vals, method='bonferroni')[1]
for i, (a, b) in enumerate(pairs):
    print(f"Adjusted p-value for {a} vs {b}: {adjusted[i]:.4f}")
```

**Interpretation of results**

- After carrying out Wilcoxon's post-hoc with Bonferroni's corrections, the p-values of all 3 pairs were < 0.05. This means that all 3 biomarkers are distinctly different from each other.

- While the 3 biomarkers are represented in different units (i.e HbA1c is shown in percentage while Fasting and OGTT glucose is shown in mmmol/L), it is still meaningful to carry out these tests as Friedman's and Wilcoxon's test are ranked-based tests.

- The results from these tests suggest that the 3 biomarkers do not behave the exact same way in individuals which is expected considering the nature of the biomarkers.

- Hence, all 3 biomarkers provides distinct information, even though they all measure metrics of glucose metabolism. These 3 biomarkers are not interchangeable.
  - Fasting glucose reflects the baseline glucose level and control, HbA1c reflects long-term glucose levels, while OGTT glucose reflects the body's ability to handle glucose intake.

- This emphasises the need to include all 3 diabetic biomarkers as dependent variables into this analysis in order to get a comprehensive understanding of the relationship between diabetes and depression.

### Comparison of depression status vs. diabetic biomarkers
- using Kruskal-Wallis test with post-hoc Dunn's test

Hypotheses to be used:

Null hypothesis (H₀): There is no difference in fasting glucose between different depression severity status.


Alternative hypothesis (Ha): At least one depression level has a significantly different fasting glucose level.

```python
# --- Classification for Depression (PHQ9_total) ---
def classify_phq9(x):
    if x >= 20:
        return "Severe"
    elif x >= 15:
        return "Moderately Severe"
    elif x >= 10:
        return "Moderate"
    elif x >= 5:
        return "Mild"
    else:
        return "None"

df_clean["Depression_Level"] = df_clean["PHQ9_total"].apply(classify_phq9)
```

```python
df_clean["Depression_Level"].value_counts()
```

```python
import scipy.stats as stats
import scikit_posthocs as sp

biomarkers = {
    "LBDGLUSI": "Fasting Glucose",
    "LBXGH": "HbA1c",
    "LBDGLTSI": "OGTT Glucose"
}

for var, label in biomarkers.items():
    print(f"\n--- Kruskal-Wallis Test: {label} ---")

    # Drop NaNs for this specific biomarker
    temp_df = df_clean[["Depression_Level", var]].dropna()

    # Prepare groups
    groups = [group[var].values for name, group in temp_df.groupby("Depression_Level")]

    # Run Kruskal-Wallis test
    stat, p = stats.kruskal(*groups)
    print(f"Statistic = {stat:.4f}, p-value = {p:.4f}")

    # Post-hoc if significant
    if p < 0.05:
        print("→ Significant. Running Dunn's post-hoc test with Bonferroni correction:")
        dunn_result = sp.posthoc_dunn(temp_df, val_col=var, group_col="Depression_Level", p_adjust="bonferroni")
        print(dunn_result)
    else:
        print("→ Not significant. No post-hoc test needed.")
```

**Interpretation of results:**

- Based on this output, we can see that there is no significant difference between diabetic biomarkers of different depression statuses
- This suggests that depression status is not a good predictor for diabetes.
- We will further explore the relationship between depression and diabetes using multiple linear regression
- To further explore the relationships, we will carry out stratified analysis using categorical variables like gender and race.

```python
for gender in df_clean["RIAGENDR"].unique():
    print(f"\n===== {gender.upper()} PARTICIPANTS =====")

    sub_df = df_clean[df_clean["RIAGENDR"] == gender]

    for var, label in biomarkers.items():
        print(f"\n--- Kruskal-Wallis Test for {label} ---")

        temp_df = sub_df[["Depression_Level", var]].dropna()
        groups = [group[var].values for name, group in temp_df.groupby("Depression_Level")]

        # Skip if not enough groups
        if len(groups) < 2:
            print("Not enough groups to compare.")
            continue

        stat, p = stats.kruskal(*groups)
        print(f"Statistic = {stat:.4f}, p-value = {p:.4f}")

        if p < 0.05:
            print("→ Significant. Running Dunn's post-hoc test:")
            dunn = sp.posthoc_dunn(temp_df, val_col=var, group_col="Depression_Level", p_adjust="bonferroni")
            print(dunn)
        else:
            print("→ Not significant.")
```

### Residual plots

Residual plots are used to assess the fit of the model.

```python
# Function to plot residuals for each model
def plot_residuals(y_var):
    X = reg_df.drop(columns=outcomes)
    y = reg_df[y_var]
    X = sm.add_constant(X)
    model = sm.OLS(y, X).fit()
    fitted_vals = model.fittedvalues
    residuals = model.resid

    plt.figure(figsize=(6, 4))
    sns.residplot(x=fitted_vals, y=residuals, lowess=True,
                  line_kws={'color': 'red', 'lw': 2})
    plt.title(f"Residual Plot for {y_var}")
    plt.xlabel("Fitted Values")
    plt.ylabel("Residuals")
    plt.axhline(0, color='black', linestyle='--', lw=1)
    plt.tight_layout()
    plt.show()

# Run for each outcome
for outcome in outcomes:
    plot_residuals(outcome)
```

### Checks for multicollinearity

The use of multiple linear regression models assumes that there is no or low multicollinearity.

This section checks for the presence of multicollinearity in our dataset to ensure that our coefficient and p-values are accurate and to ensure we can form correct interpretations from the models.

To check for multicollinearity, the variance inflation factor for each variable is used.

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor

# Create feature matrix (no outcomes, with intercept)
X = reg_df.drop(columns=outcomes)
X = sm.add_constant(X)

# Calculate VIF for each feature
vif_data = pd.DataFrame()
vif_data["Variable"] = X.columns
vif_data["VIF"] = [variance_inflation_factor(X.values, i) for i in range(X.shape[1])]

vif_data
```

**Interpretation of results:**

VIF values of less than 5 indicate that there is low or no multicollinearity in our regression models.

This confirms that the p-values and coefficients generated from the regression models are accurate and can be used for interpretation.

# 6. Conclusions and future work




Although previous literature suggests a possible link between depression and diabetes, our findings show a more nuanced relationship.

Descriptive and stratified analyses showed no consistent significant differences in glycaemic biomarkers across PHQ-9 scores. Even in subgroup analysis of gender and ethnicity, most comparisons were not statistically significant, suggesting that depression severity alone may not be a strong independent predictor of glycaemic outcomes.

However, through linear regression models, we found several consistent predictors of poorer glycaemic control. Insulin resistance and age were shown as the strongest continuous predictors across all three biomarkers, while race also showed notable associations, indicating the influence of both biological and socio-cultural factors.


PHQ-9 scores remained non-significant across all models, even after adjusting for confounders. An unexpected finding was that participants reporting less than an hour of sedentary activity daily had higher glucose levels than those reporting 1 hour per day. This result could be due to small sample size or unmeasured cofounding.

Model performance suggested that while linear models explained some variability, they may not fully capture the complexity of the relationships.

All in all, these findings collectively reinforce the significance of traditional metabolic predictors such as insulin resistance and age in glycaemic control. It also questions the predictive value of depression severity on its own in a general adult population.


From a public health perspective, this suggests that although integrated care for depression and diabetes is important, depression alone may not serve as a reliable predictor for glycaemic control in population-level screening. Therefore, screening and prevention programmes should continue to prioritise well-established metabolic risk factors while recognising that depression still plays an indirect role in metabolic health.

Importantly, certain subgroups may experience stronger or more complex links between depression and glycaemic control. For instance, older adults and women may experience different behavioral pathways linking depressive symptoms to glucose dysregulation (Cai et al., 2023). Public health strategies should therefore be sensitive to these subgroup-specific risks and consider tailored interventions rather than a one-size-fits-all approach.


While this study provided valuable insights into the relationship between depression severity and glycaemic biomarkers, several avenues remain for future exploration.

First, future research could use logistic regression models to evaluate the odds of prediabetes or diabetes across different levels of depression severity. This may provide clearer clinical relevance compared to continuous outcomes alone, especially when identifying high-risk individuals in screening programmes (Cai et al., 2024). Such analysis could also includef interaction terms to see whether the link between depression and glycaemic markers varies by demographic or lifestyle factors in order to improve the precision of risk stratification.

Additionally, temporal trend analysis using more NHANES cycles or newer datasets could help assess whether these relationshops have evolved over time due to changing public health dynamics. This could show us changes in the strength or directions of links and highlight growing at-risk groups in response to societal changes, such as sedentary lifestyles or increased anti-depressant usage.

Further studies could also benefit from including additional predictors such as use antidepressants and chronic stress. This is because these predictors may influence both mood and glucose regulation (Cai et al., 2024). Studies could also help uncover indirect pathways between depression and glycaemic control. Through integration of psychological, biological and behavioural factors into predictive models, researchers can gain a better understaning on the association of depression with mental health.

We did not use these factors in our investigation but they can provide a more complete understanding of the relationship between mental and metabolic health.

Finally, this project could be extended into a publication by refining the current models, validating findings with alternative datasets or focusing on specific groups. Such efforts could help develop targeted preventive strategies and inform integrated care approached for patients with overlapping mental and metabolic conditions.
